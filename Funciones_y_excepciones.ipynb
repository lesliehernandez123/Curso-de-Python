## Funciones que llaman a otras funciones


```python
def saludar(nombre):
  return f"Hola, {nombre}!"

def despedirse(nombre):
  return f"Adiós, {nombre}!"

def mensaje(nombre):
  saludo = saludar(nombre)
  despedida = despedirse(nombre)
  return saludo + "\n" + despedida

print(mensaje("Leslie Hernández"))
```

```python
# Ahora supongamos que tenemos una heladería
def pedir_sabor():
  sabores = ["chocolate", "fresa", "vainilla"]
  print("Sabor disponible", ",", sabores)

  # strip() elimina los espacios tanto antes como después y lower() hace minúsculas las palabras
  sabor = input("Elige un sabor: ").strip().lower()

  while sabor not in sabores:
    print("No tenemos ese sabor")
    sabor = input("Elige un sabor: ").strip().lower()

  return sabor

def pedir_cantidad():
  cantidad = int(input("Cuántas porciones quieres? "))
  return cantidad
  while cantidad <=0:
    cantidad = int(input("Cuántas porciones quieres? "))
  return cantidad

#Función que llama a las demás
def pedir_helado():
  sabor = pedir_sabor()
  cantidad = pedir_cantidad()
  print(f"Pedido: {cantidad} porciones de {sabor}")
  return sabor, cantidad

total_pedidos = int(input("Cuántas ordenes quieres hacer? "))
for i in range(total_pedidos):
  pedir_helado()
```

## **Funciones Lambda**
Una función Lambda es una función acortada para escribir una función de esta forma:

nombreFunción = lambda parámetros: acción

La diferencia entre las funciones lambda y las funciones normales es que suelen ser un poco más limitadas.

```python
sumar = lambda x, y: x + y
print(sumar(2,3))
```

```python
descuento = lambda p, d: (p - (p * d / 100), (p * d / 100))
nuevoPrecio = descuento(500, 40)
print("El precio a pagar es", nuevoPrecio[0], ", descuento de: ", nuevoPrecio[1])
```

Tenemos otros tipos de errores, por ejemplo, el de **ValueError** que sale al ingresar el tipo incorrecto de dato
Digamos que un usuario ingresa texto en lugar de un número, este error nos ayuda a manejarlo

```python
try:
  numero = int(input("Ingresa un número: "))
  print(numero)
except ValueError:
  print("No es un número")
except ZeroDivisionError:
  print("No se puede dividir por cero")
```

También podemos manejar más de una excepción, por ejemplo, con la decisión podemos tener tanto el tipo de valor como el de la división entre 0

```python
try:
  dividendo = float(input("Ingresa el dividendo: "))
  divisor = float(input("Ingresa el divisor: "))
  division = dividendo / divisor
except ZeroDivisionError:
  print("No se puede dividir por cero")
except ValueError:
  print("Error: ingrese un valor numérico")
```

### **Actividad**: Obtener el promedio del grupo e identificar el promedio más alto

```python
def promediosYmayor(promedios):
  promedio = 0
  for x in promedios:
    promedio += x
  promedio = promedio / len(promedios)
  return promedio, max(promedios)

promedios = []
continuar = ""

while True:
    if len(promedios) > 0:
      continuar = ""
      while continuar not in ("si", "no"):
        continuar = input("Ingrese una respuesta válida (si/no): ").lower()
      if continuar == "no":
        break

    while True:
      try:
        calif = int(input("Indique la calificación: "))
        if calif < 0 or calif > 100:
          print("Ingrese una calificación válida")
        else:
          promedios.append(calif)
          break
      except ValueError:
        print("Ingrese una calificación válida")

prom, mayor = promediosYmayor(promedios)
print("\nResultados:")
print(f"El promedio del grupo es: {prom}")
print(f"La calificación más alta del grupo es: {mayor}")
```

## **Actividad**: Calculadora

```python
# Calculadora -> suma, resta, multiplicación, división y potencia
def suma (a,b):
  return a + b
def resta (a,b):
  return a - b
def multiplicación (a,b):
  return a * b
def división (a,b):
  if b == 0:
    return "No se puede dividir por cero"
  return a / b
def potencia (a,b):
  return a ** b

def menu():
  print("------------ MI CALCULADORA --------------")
  print("1. Suma")
  print("2. Resta")
  print("3. Multiplicación")
  print("4. División")
  print("5. Potencia")
  print("6. Salir")

while True:
  menu()
  try:
    op = int(input("Selecciona una operación (1-6): "))
  except ValueError:
    print("Ingrese un número válido")
    continue

  if op == 6:
    print("SALIENDO...")
    break

  if op < 1 or op > 6:
    print("Ingrese una opción válida")
    continue

  try:
    a=float(input("Ingrese el primer número: "))
    b=float(input("Ingrese el segundo número: "))
  except ValueError:
    print("Ingrese un número válido")
    continue

  if op == 1:
    print("La suma es: ", suma(a,b))
  elif op == 2:
    print("La resta es: ", resta(a,b))
  elif op == 3:
    print("La multiplicación es: ", multiplicación(a,b))
  elif op == 4:
    print("La división es: ", división(a,b))
  elif op == 5:
    print("La potencia es: ", potencia(a,b))
```
