## Estructuras de Datos

Las estructuras de datos son formas de organizar información para trabajar con ellas de manera eficiente.

- Listas -> ordenadas, modificadas, permiten duplicados
- Tuplas -> ordenadas, inmutables, permiten duplicados
- Sets -> no son ordenados, sin duplicados
- Diccionarios -> pares por clave:valor

Nota:
- Pilas (FILO -> First In, Last Out)
- Colas/filas (FIFO -> First In, First Out)
- Listas enlazadas o doblemente enlazadas

## **Listas**
Las listas son colecciones ordenadas que podemos modificar y almacenan elementos de distintos tipos de datos

```python
lista_de_compras = ["manzanas","pan","leche", "huevos", "mantequilla","queso","jamon"]
print(lista_de_compras)
print(lista_de_compras[4])
```

```python
# Agregar un elemento a la lista .append()
lista_de_compras.append("sopa")
print(lista_de_compras)
```

```python
# Agregar un elemento en una posición específica
lista_de_compras.insert(2,"arroz")
print(lista_de_compras)
```

```python
# Para eliminar un elemento de una lista existe el "remove" y el "pop"
lista_de_compras.remove("sopa")
print(lista_de_compras)
```

```python
lista_de_compras.pop() # Sin nada dentro del parantesis borra el último elemento
lista_de_compras.pop(7) # o puedes colocar la posición
#Nota: te imprime el valor eliminado
```

```python
# Podemos recorrer el arreglo por medio de un for
for elementos in lista_de_compras:
  print(elementos)
```

```python
# len() obtiene el tamaño de la lista
print(len(lista_de_compras))
for i in range(len(lista_de_compras)):
  print(f"{i + 1} - {lista_de_compras[i]}")
```

```python
# Piedra, papel o tijera
import random
opciones = ["piedra", "papel", "tijeras"]
while True:
  while True:
    eleccion = input("\nElige piedra, papel o tijeras (o escribe salir si ya no quieres jugar): ")
    if eleccion not in opciones and not eleccion == "salir":
      print("Opción no válida, intenta de nuevo.")
    else:
      break

  if eleccion == "salir":
    print("¡Gracias por jugar!")
  break
aleatoria = random.choice(opciones)
print("La computadora eligió:", aleatoria)

# Determinamos el ganador
if eleccion == aleatoria:print("Empate")
elif (eleccion == "piedra" and aleatoria == "tijeras") \
    or (eleccion == "papel" and aleatoria == "piedra") \
    or (eleccion == "tijeras" and aleatoria == "papel"):
  print("¡Ganaste!:))")
else:
  print("Perdiste:((")
```

```python
# Para generar matrices
calif = [[100,88,93], [99,56,72]]
i = 1
for cali_i in calif:
  print(f"\nAlumno {i}")
  j = 1
  for cali_j in cali_i:
    print(f"Calificación {j}: {cali_j}")
    j += 1
  j = 0
  i += 1
```

## **Sets**
Los sets son colecciones SIN orden y que NO permiten elementos duplicados

```python
nombre = {"Angel", "Fernando", "Jose", "Paulina","Jose"} # lo borra directamente el duplicado
print(nombre)
print(type(nombre))
```

```python
# Add
nombre.add("Leslie")
nombre
```

```python
# Unión (.union o |)
matutino = {20322,22183,45332,12345}
vespertino = {20322, 11111, 22222, 33333}
total_matriculas = matutino.union(vespertino)
print(total_matriculas)

total_matriculas = matutino | vespertino
print(total_matriculas)
```

```python
# Intersección
ca = {1234,2345,3456,2873,39483948,39483904,23892832}
al = {1234,2345,454535,3498394}
ca_al = ca.intersection(al)
print (ca_al)

ca_al = ca & al
print (ca_al)
```

```python
# Diferencia
solo_calculo = ca.difference(al)
print(solo_calculo)

solo_calculo = ca - al
print(solo_calculo)
```

```python
nombre = {"Angel", "Fernando", "Jose", "Paulina","Jose"}
#remove, pop
nombre.remove("Jose") # Puedes eliminar el que tu quieras
print(nombre)
nombre.pop() # Elimina un elemento al azar
nombre
```

```python
# Eliminar todo el set
nombre.clear()
nombre
```

## Diccionarios
Los diccionarios son colecciones no ordenadas, de pares clave:valor

```python
alumno = {
    #clave     valor
    "nombre": "Leslie Hernández",
    "edad": 21,
    "carrera": "Ingeniería Química",
    "semestre": 10,
    "promedio": 88,
    "calificaciones":[100,94,88,70,100]}
print(type(alumno))
alumno
```

```python
print("El alumno ", alumno["nombre"]," tiene ",alumno["edad"]," años", " esta estudiando", alumno["carrera"])
```

```python
# Agregar al diccionario
alumno["area preferida"] = "Procesos de Separación"
alumno
```

```python
for valor in alumno:
  print(alumno[valor])
```

```python
for aspecto, valor in alumno.items():
  print(f"{aspecto}: {valor}") # devuelve una vista de todos los pares de datos clave:valor
```

```python
# Para borrar los valores se usa .clear()
alumno.clear()
alumno
```

```python
#Tenemos el .get que nos ayuda a obtener el valor (value) de cierta clave (key)
print(alumno.get("nombre"))
```

```python
alumno_1 = alumno.items()
print(alumno_1)
print(type(alumno_1))

# También se puede convertir a lista
alumno_lista = list(alumno_1)
print(alumno_lista)
print(type(alumno_lista))
```

```python
# El método .values() sirve para obtener los valores del diccionario, también se puede usar en la interacción
alumno_v = alumno.values()
print(alumno_v)
```

```python
if "carrera" in alumno:
  print("Existe carrera")
else:
  print("No existe carrera")

if ("dinero") in alumno:
  print("Alumno tiene dinero")
else:
  print("Alumno no tiene dinero")
```

```python
# .copy() puedes copiar todo el diccionario en una nueva variable
alumno_2 = alumno.copy()
alumno_2

#Update -> Sirve para cambiar los valores (actualizar)
alumno.update({"semestre":9})
alumno
```

Actividad: **Supermercado**

```python
# Lista de productos
productos = ["manzana", "banana", "leche", "pan"]

# Tupla de precios
precios = (10, 5,20, 15)

# Diccionario de inventario: producto -> {"precio": x, "stock": y}
inventario = {productos [i]: {"precio": precios[i], "stock": 10} for i in range(len(productos))}
# Clave: nombre del producto
# Valor: otro diccionario con "precio" y "stock"

# Set de productos en oferta
ofertas = {"manzana", "pan"}

# Mostrar productos
for p in productos:
  msg = "OFERTA!!!" if p in ofertas else "Grandes precios!"
  info = inventario[p]
  print(f"{p.title()} - ${info['precio']} - Stock: {info['stock']} -> {msg}")

# Compra simple
compra = input("\n¿Qué productos deseas comprar?").strip().lower()
if compra in inventario and inventario[compra]["stock"]>0:
  inventario[compra]["stock"] -= 1
  print(f"Compraste 1 {compra}. Stock restante: {inventario[compra]['stock']}")
else:
  print("Producto no disponible")
```
