**Actividad**: Calculadora (con clases)

```python
class Calculadora:
  def __init__(self, num1, num2):
    self.num1 = num1
    self.num2 = num2
  def sumar(self):
    print(self.num1 + self.num2)
  def restar(self):
    print(self.num1 - self.num2)
  def multiplicar(self):
    print(self.num1 * self.num2)
  def dividir(self):
    if self.num2 == 0:
      print("No se puede dividir entre cero")
    else:
      print(self.num1 / self.num2)

calculadora = Calculadora(10, 0)
calculadora.sumar()
calculadora.restar()
calculadora.multiplicar()
calculadora.dividir()
```

## **Encapsulamiento** y propiedades

```python
class Cuenta:
  def __init__(self,saldo):
    self.__saldo = saldo # Privada
# Sirve para proteger los datos internos de un objeto y controlar como se leen o modifican.
cuenta = Cuenta(1000)
cuenta.Cuenta = 2000 # si puedes editarlo
print(cuenta.__saldo) # Error no se puede acceder directamente
```

```python
# Propiedades: Getters y Setters
class Cuenta:
  def __init__(self,saldo):
    self.__saldo = saldo # Privada
  @property # Getter
  def saldo(self):
    # Permite leer el valor de __saldo usando c.saldo
    return self.__saldo
  @saldo.setter # Setter
  def saldo(self, valor): # Nota: si se declara un getter se debe poner un setter para editarlo
    # Se ejecuta cuando hacemos c.saldo = valor
    self.__saldo = valor

c = Cuenta(300)
print(c.saldo)
c.saldo = 4000
print(c.saldo)
```

## **Herencia y Abstracción**

```python
# Herencia

class Vehículo: # Clase Base (padre)
  def __init__(self, marca, modelo):
    self.marca = marca
    self.modelo = modelo

  def descripción(self):
    return f"{self.marca} : {self.modelo}"
  def mover(self):
    return "El vehículo se está moviendo..."

# ------------------------------------------------------------------------------

class Coche(Vehículo): # subclase 1 (hija)
  def mover(self):
    return "El coche se está condunciendo"

# ------------------------------------------------------------------------------

class Moto(Vehículo): # subclase 2 (hija)
  def mover(self):
    return "La moto avanza sobre dos ruedas"

# ------------------------------------------------------------------------------

coche2 = Vehículo("BYD", "Dolphin") # objetos
coche = Coche("Toyota", "Corolla")
moto = Moto("Yamaha", "R6")

print(coche2.descripción())
print(coche2.mover())
print(coche.descripción())
# aunque no volvimos a declarar el método dentro de descripción, podemos seguir usandolo ya que lo heredó del padre

print(coche.mover())
print(moto.descripción())
print(moto.mover())
```

```python
class Persona:
  def __init__(self, nombre,edad):
    self.nombre = nombre
    self.edad = edad
  def presentarse(self):
    print(f"Hola, soy {self.nombre} y tengo {self.edad} años")

class Estudiante(Persona):
  def __init__(self, nombre, edad, carrera): # no está bien tener 2 variables
    super().__init__(nombre, edad) # función super() para no crear variables repetidas
    self.carrera = carrera
  def presentarse(self):
    super().presentarse()
    print(f"Estoy estudiando la carrera de {self.carrera}")
# Reutiliza el constructos de la clase padre, lo cual evita duplicar el código
# Ejecuta el comportamiento base y luego lo extiende

est = Estudiante("Leslie", 21, "Ingeniería Química")
est.presentarse()
```

## **Polimorfismo**
El polimorfismo significa que diferentes objetos pueden responder al mismo mensaje (método) de forma distinta, siempre y cuando compartan una interfaz en común.

***Composición HAS-A***
- ¿Por qué se llama composición? R= Porque estás componiendo (armando) un objeto grande usando otros objetos
- ¿Qué es? Es un principio de POO donde: una clase NO hereda de otra, si no que contiene un objeto de esa otra clase.
- Ej: una persona no puede ser un banco pero si puede tener una cuenta en el banco

```python
class Banco:
  def __init__(self, dinero):
    self.dinero = dinero
  def ver_total(self):
    print(f"Total en la cuenta: ${self.dinero}")

class Persona:
  def __init__(self, cuenta: Banco): # Indica que este parámetro debería ser una instancia de la clase Banco
    self.cuenta = cuenta # Persona TIENE una cuenta de banco, no hereda de el
  def depositar(self, monto):
    self.cuenta.dinero += monto
    self.cuenta.ver_total()
```

## **Composición vs Herencia**
No significa que alguna sea mejor que otra, simplemente depende de lo que busques, hay casos donde una conviene más que otra

- **Herencia: un gato ES un animal** (NO "Un gato TIENE un animal")
- **Composición: una persona TIENE una cuenta** (NO "Una persona ES una cuenta)

### **Ejercicio integrador**:
Se quiere modelar un pequeño sistema para vehículo (coche, moto, bicicleta). Todos comparten características básicas, pero cada uno tiene un motor distinto o no lo tiene, por lo tanto, debes manejarlo como "Composición".
*Opcional: usa polimorfismo*

```python
class Vehículo:
  def __init__(self, marca):
    self.marca = marca

  def mover(self):
    raise NotImplementedError("Este método DEBE ser implementado en las subclases")

# ------------------------------------------------------------------------------

class Motor:
  def __init__(self, caballos):
    self.caballos = caballos

  def potencia(self):
    return f"{self.caballos} HP"

# ------------------------------------------------------------------------------

class Coche(Vehículo):
  def __init__(self, marca, caballos):
    super().__init__(marca)
    self.motor = Motor(caballos) # Composición

  def mover(self): # Polimorfismo
    return f"El coche {self.marca} avanza con motor de {self.motor.potencia()}"

# ------------------------------------------------------------------------------

class Moto(Vehículo): # Polimorfismo
  def __init__(self, marca, caballos):
    super().__init__(marca)
    self.motor = Motor(caballos) # Composición

  def mover(self):
    return f"La moto {self.marca} acelera con motor de {self.motor.potencia()}"

# ------------------------------------------------------------------------------

class Bicicleta(Vehículo):
  def mover(self): # Polimorfismo
    return f"La bicicleta {self.marca} avanza con pedaleo humano"

# ------------------------------------------------------------------------------

def probar_vehículos(lista): # Llamada al polimorfismo
  for v in lista:
    print(v.mover())

# La usaremos en esta ocasión en una lista, SI se puede hacer listas de clases.
vehículos = [
    Coche("Toyota", 150),
    Moto("Yamaha", 90),
    Bicicleta("Trek")
]

probar_vehículos(vehículos)
```
