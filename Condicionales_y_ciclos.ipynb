## **elif**

Ahora hagamos un programa en el que preguntamos la calificación del 0 al 100 y queremos saber su equivalente en el sistema de evaluación estadounidense. Para simplicidad, basemonos en esta:

- A+ = 100
- A = 90 - 99
- B = 80 - 89
- C = 70 -79
- D = 60 - 69
- F = < 60

```python
cal = float(input("Ingresa tu calificación: "))
if cal == 100:
  print("Obtuviste una A+")
elif cal >= 90:
    print("Obtuviste una A")
elif cal >= 80:
    print("Obtuviste una B")
elif cal >= 70:
    print("Obtuviste una C")
elif cal >= 60:
    print("Obtuviste una D")
else:
  print("Obtuviste una F")
```

## Operaciones lógicas con condicionales

```python
# and, or, not
cal = float(input("Ingresa tu calificación: "))
if cal >= 0 and cal <= 100:
  if cal == 100:
    print("Obtuviste una A+")
  elif cal >= 90:
      print("Obtuviste una A")
  elif cal >= 80:
      print("Obtuviste una B")
  elif cal >= 70:
      print("Obtuviste una C")
  elif cal >= 60:
      print("Obtuviste una D")
  else:
    print("Obtuviste una F")
else:
  print("Error: La calificación debe estar entre 0 y 100")
```

```python
# El modulo (%) es un operador matemático que devuelve el residuo de una división
# Ejemplo: 10 / 3 = 3, y sobra 1
num = 9
if num % 3 == 0 or num % 5 == 0:
  print(f"El número {num} es divisible por 3 o 5")

#saber si el número es par o impar usando not
num = 6
if not num % 2 == 0:
  print(f"El número {num} es impar")
else:
  print(f"El número {num} es par")
```

## **ciclo for**
Los llamados ciclos for, nos ayudan a realizar acciones durante una cantidad determinada de veces.

```python
# Imprimir los números del 0 al 4
for i in range(5):
  print(f"Valor de i: {i}")

print ("")

for i in range(2,5):
  print(i)
```

```python
cadena = "Python"
for letras in cadena:
  print(letras)
```

```python
nombre = "Leslie"
for i in nombre:
  print(i)
```

```python
rango = 2,5,7,9,10,15,20
for i in rango:
  print(i)
```

```python
for i in range(2,10,2):
  print(i)
```

## **While**

```python
suma = 0
numeros = int(input("Ingresa los números a sumar: "))

while numeros != 0: # !=distinto de 0
  suma += numeros
  numeros = int(input("Ingresa los números a sumar o 0 para salir: "))

print("")
print(f"La suma de los números es: {suma}")
```

```python
# Usar while como ciclo for
numero = 0
while numero <= 12:
  print (numero)
  numero += 1
```

Actividad: adivina el número

```python
print("Juego: Adivina el número")
numero_secreto = int(input("Ingresa un número entre 1 y 20: "))
intentos = 0
max_intentos = 4

while intentos < max_intentos:
  propuesta = int(input("Adivina el número: "))
  intentos += 1
  if propuesta == numero_secreto:
    print("¡Correcto!")
    break
  else:
    print("Incorrecto, intenta de nuevo.")

# Opcional
if intentos == max_intentos and propuesta != numero_secreto:
  print("Se acabaron los intentos")
print("El número era:", numero_secreto)
```
