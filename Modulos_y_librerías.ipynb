## ¿Qué es un módulo?
Un **modulo** es una unidad de organización del código que permite encapsular y dividir el software en partes más manejables, generalmente representado como un archivo que contiene código relacionado como: funciones, variables (constantes) o métodos de clase.

### **JSON**

Existe 1 solo tema que TODO programador independientemente del área DEBE de saber
**JSON** es un formato para guardar y transmitir datos. Es muy útil en APIs, archivos de configuración y almacenamiento de información.
(Se parece a un diccionario de Python, pero es solo texto)

¿Para qué sirve?
  - Para convertir objetos de Python -> Texto JSON y también texto JSON -> objetos de Python
  
  Esto se llama:
  - Serializar: Convertir un objeto de Python en texto JSON
  - Deserializar: Convertir un texto JSON a un objeto usable de Python

```python
# Funciones principales:
# 1) json.dums(obj): Convierte un diccionario u objeto de Python en un string JSON
import json
obj = {"nombre":"Leslie"
, "edad": 25}
print(type(obj))
s = json.dumps(obj, ensure_ascii= False)
print(s)
print(type(s))
```

```python
# json.loads(texto_json): Convierte un string JSON en un diccionario Python
restaurar = json.loads(s)
print(restaurar)
print(type(restaurar))
# Al venir de una página/persona/archivo desconocido
'''
Pueden contener:
Datos malinformados
Información inesperada
Campos faltantes
Datos demasiado grandes
Variables incorrectas que rompan tu programa
'''
```

## **Numpy**
Provee el tipo de dato ndarray, operaciones vectorizadas, broadcasting, utilidades de algebra lineal y funciones matemáticas en arrays. Es la base de muchas librerías científicas en Python.

***Nota importante:*** *El arreglo de Numpy no es igual a la lista de Python, este NO puede tener diferentes tipos de datos en el mismo.*

```python
#pip install numpy cuando lo usas en Python, arreglo matricial
import numpy as np
A = np.array([[1,2,3],[4,5,6]]) # array (2,3)
v = np.array([10,20,30]) # array (1,3)
print(f"Suma de matrices:", A + v)
print("")
print("-----------------------")
print("")
prueba = [1,2,3]
prueba2 = [1,2,3]
print(f"Si lo hicieramos como lista: ", prueba + prueba2) # unión
```

```python
# Más funciones de Numpy
z = np.zeros((3,3)) # array (3,3)
print(z)
o = np.ones((2,4))
print(o)
r = np.random.rand(3,3) #arreglo random de (3,3)
print(r)
```

```python
# Rangos numéricos
nums = np.arange(0,10,2) # valores de 0 a 10 de 2 en 2 sin contar el 10
print(nums)
x = np.linspace(0,10,5) # 5 valores igualmente espaciados entre 0 y 10
print(x)
```

```python
# Operaciones con matrices
a = np.array([1,2,3,4,5])
b = np.array([6,7,8,9,10])
print(a + b) # suma
print(a - b) # resta
print(a * b) # multiplicación
print(a / b) # división
print(a ** b) # potencia
```

```python
print(a.dot(b)) # producto de dos matrices
c = np.array([[1,2,3],[4,5,6]])
d = np.array([[10,20],[30,40],[50,60]])
''' 1*10 + 2*30 + 3*50'''
print(c.dot(d))
```

```python
data = np.array([1,2,3,4,5])
print(data.min()) # mínimo
print(data.max()) # máximo
print(data.sum()) # suma
print(data.mean()) # promedio
print(data.std()) # desviación estándar
```

```python
A = np.array([1,6,3,9,2])
print(A[A>5])
```

## **Matplotlib**
Sirve para graficar y se le puede dar formato deseado.

```python
import matplotlib.pyplot as plt
x = [1,2,3,4,5]
y = [2,3,5,7,11]
# linestyle, linewidth
plt.plot(x,y, marker= "o") # Gráfica de líneas
plt.title("Título")
plt.xlabel("Eje x")
plt.ylabel("Eje y")
plt.grid(False)
plt.show()
```
