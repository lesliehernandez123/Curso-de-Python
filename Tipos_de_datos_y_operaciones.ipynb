## Operaciones matemáticas

```python
# Suma ordinaria
print("Suma: 3 + 4")
leslie = 3
print(leslie + 4)
x1,x2 = 20.5,40
print("x1 + x2")
print(x1 + x2)
print("")

# Resta ordinaria
print("Resta: x2 - x1")
print(x2 - x1)
print("")

# Multiplicación ordinaria
L1,L2,L3,L4,L5 = 1,2,3,4,5
print("Multiplicación: L1*L2*L3*L4*L5")
print(L1*L2*L3*L4*L5)
print("")

# División ordinaria
print("División: L5 / L2")
print(L5 / L2)
print("")

# División de piso
print("División de piso: L5 // L2")
print(L5 // L2)
print("")

# Potencia
print("Potencia: L2 ^ L3")
print(L2 ** L3)
print("")

#Raiz
print("Raíz: L4 ^ (1/2)")
print(L4 ** (1/2))
print("")
```

```python
# Área de un círculo
pi = 3.14
r = 4
A = pi * r ** 2
print("El área del círculo es: ", A)
```
